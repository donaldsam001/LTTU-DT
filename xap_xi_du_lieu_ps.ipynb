```python
import sympy as sp
```

```python
class QuadraticOptimizer:

    def __init__(self, X, Y):

        # Chuyển dữ liệu sang phân số chính xác
        self.X = sp.Matrix([
            [sp.Rational(str(value)) for value in row]
            for row in X
        ])

        self.Y = sp.Matrix([
            [sp.Rational(str(value))] for value in Y
        ])

        self.n, self.m = self.X.shape
        self.k = self.m + 1

        # 1. Xây dựng ma trận Q
        self.Q = sp.ones(self.n, 1).row_join(self.X)

        # 2. Tính A, b, c
        self.A = self.Q.T * self.Q
        self.b = -self.Y.T * self.Q
        self.c = (self.Y.T * self.Y)[0]

        # 3. Khởi tạo P
        self.P = sp.Matrix(
            sp.symbols(f'p0:{self.k}')
        )

        # Tính từng thành phần dưới dạng số vô hướng
        quadratic = (self.P.T * self.A * self.P)[0]

        linear = (self.b * self.P)[0]

        constant = self.c

        # Xây dựng hàm mục tiêu
        self.f = sp.expand(
            sp.Rational(1, 2) * quadratic
            + linear
            + sp.Rational(1, 2) * constant
        )

        # 5. Gradient
        self.gradient = sp.Matrix([
            sp.diff(self.f, p)
            for p in self.P
        ])

        # 6. Hessian
        self.hessian = sp.hessian(
            self.f, list(self.P)
        )

    # Hiển thị A, b, c
    def show_ABC(self):

        print("\n========== MATRIX A ==========")
        sp.pprint(self.A)

        print("\n========== VECTOR b ==========")
        sp.pprint(self.b)

        print("\n========== CONSTANT c ==========")
        sp.pprint(self.c)

    # Hiển thị hàm mục tiêu
    def show_function(self):

        print("\n========== OBJECTIVE FUNCTION ==========")
        sp.pprint(self.f)

    # Hiển thị Gradient
    def show_gradient(self):

        print("\n========== GRADIENT ==========")
        sp.pprint(self.gradient)

    # Hiển thị Hessian
    def show_hessian(self):

        print("\n========== HESSIAN ==========")
        sp.pprint(self.hessian)

    # Tính định thức các ma trận con chính đầu
    def show_determinants(self):

        print("\n========== HESSIAN DETERMINANTS ==========")

        for i in range(1, self.k + 1):

            sub_matrix = self.hessian[:i, :i]

            determinant = sp.factor(sub_matrix.det())

            print(f"\nD{i} =")
            sp.pprint(determinant)

    # Kiểm tra Hessian
    def check_positive_definite(self):

        print("\n========== CHECK HESSIAN ==========")

        determinants = []

        for i in range(1, self.k + 1):

            determinant = self.hessian[:i, :i].det()
            determinants.append(determinant)

        if all(d > 0 for d in determinants):

            print("Hessian is POSITIVE DEFINITE")
            print("=> Unique global minimum")

        else:
            print("Hessian is NOT positive definite")

    # Tìm nghiệm tối ưu
    def solve(self):

        print("\n========== OPTIMAL SOLUTION ==========")

        # Giải hệ A.P = -b.T bằng phân số chính xác
        P_opt = self.A.inv() * (-self.b.T)

        # Hiển thị nghiệm
        for i, value in enumerate(P_opt):

            print(f"p{i} = ", end="")
            sp.pprint(value)

        quadratic = (P_opt.T * self.A * P_opt)[0]

        linear = (self.b * P_opt)[0]

        f_min = sp.simplify(
            sp.Rational(1, 2) * quadratic
            + linear
            + sp.Rational(1, 2) * self.c
        )

        print("\nf_min =")
        sp.pprint(f_min)

        print("\nf_min =")
        sp.pprint(f_min)

        # Gradient tại nghiệm
        gradient_opt = self.gradient.subs(
            dict(zip(self.P, list(P_opt)))
        )

        print("\nGradient at optimum:")
        sp.pprint(gradient_opt)

        return P_opt
```

```python
# ==========================================
# MAIN
# ==========================================

if __name__ == "__main__":

    # Dữ liệu đề bài
    X = [
        [30],
        [32],
        [34],
        [36],
        [38],
    ]

    Y = [
        36.2,
        36.2,
        37.8,
        39.7,
        43.7,
    ]

    optimizer = QuadraticOptimizer(X, Y)

    optimizer.show_ABC()

    optimizer.show_function()

    optimizer.show_gradient()

    optimizer.show_hessian()

    optimizer.show_determinants()

    optimizer.check_positive_definite()

    optimizer.solve()
```
